# The original messy vectors

```python
r1 = np.array([3,0,3,0])
r2 = np.array([3,2,3,3])
r3 = np.array([-7,-4,-5,-6])
r4 = np.array([-7,-3,-11,2])
```

```python
# Process the first vector r1 and create a nice vector u1

# Select a vector
t = r1

# Subtract off how much that vector points in the nice directions
n = t

# Find the size of n:
np.sqrt(n.dot(n))

# Create the nice new u vector:
u1 = n/np.sqrt(n.dot(n))
print(u1)
```

```python
# Let's check it has size 1 and is orthogonal to our other nice vectors

print("The size of our new vector is", u1.dot(u1))

print("Is the vector orthogonal to our existing u's?")
```

```python
# Process the first vector r2 and create a nice vector u2

# Select a vector
t = r2

# Subtract off how much that vector points in the nice directions
n = t - t.dot(u1)*u1

# Find the size of n:
np.sqrt(n.dot(n))

# Create the nice new u vector:
u2 = n/np.sqrt(n.dot(n))
print(u2)
```

```python
# Let's check it has size 1 and is orthogonal to our other nice vectors

print("The size of our new vector is", u2.dot(u2))

print("Is the vector orthogonal to our existing u's?")
print(u2.dot(u1))
```

```python
# Process the first vector r3 and create a nice vector u3

# Select a vector
t = r3

# Subtract off how much that vector points in the nice directions
n = t - t.dot(u1)*u1 - t.dot(u2)*u2

# Find the size of n:
np.sqrt(n.dot(n))

# Create the nice new u vector:
u3 = n/np.sqrt(n.dot(n))
print(u3)
```

```python
# Let's check it has size 1 and is orthogonal to our other nice vectors

print("The size of our new vector is", u3.dot(u3))

print("Is the vector orthogonal to our existing u's?")
print(u3.dot(u1))
print(u3.dot(u2))
```

```python
# Process the first vector r3 and create a nice vector u3

# Select a vector
t = r4

# Subtract off how much that vector points in the nice directions
n = t - t.dot(u1)*u1 - t.dot(u2)*u2 - t.dot(u3)*u3

# Find the size of n:
np.sqrt(n.dot(n))

# Create the nice new u vector:
u4 = n/np.sqrt(n.dot(n))
print(u4)
```

```python
# Let's check it has size 1 and is orthogonal to our other nice vectors

print("The size of our new vector is", u4.dot(u4))

print("Is the vector orthogonal to our existing u's?")
print(u4.dot(u1))
print(u4.dot(u2))
print(u4.dot(u3))
```
